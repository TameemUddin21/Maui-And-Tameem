```python
import numpy as np
def create_matrix(rows, cols):
    return np.random.rand(rows, cols)

def matrix_multiply(A, B):
    return np.dot(A, B)

def compute_statistics(matrix):
    return {
        "mean": np.mean(matrix),
        "std_dev": np.std(matrix),
        "min": np.min(matrix),
        "max": np.max(matrix),
        "sum": np.sum(matrix)
    }

def main():
    print("NumPy Dummy Project")

    A = create_matrix(3, 3)
    B = create_matrix(3, 3)

    print("\nMatrix A:")
    print(A)

    print("\nMatrix B:")
    print(B)

    C = matrix_multiply(A, B)

    print("\nA @ B:")
    print(C)

    stats = compute_statistics(C)

    print("\nStatistics of Result:")
    for key, value in stats.items():
        print(f"{key}: {value}")


if __name__ == "__main__":
    main()
```
